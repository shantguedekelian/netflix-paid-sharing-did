```python
# Find countries that appear more than once
dupes = raw_df[raw_df.duplicated(subset="country", keep=False)].sort_values("country")
print("Countries appearing in multiple waves (keeping earliest date):\n")
print(dupes[["country", "wave", "rollout_date"]].to_string(index=False))

# Keep the row with the earliest rollout_date for each country
treatment_df = (
    raw_df.sort_values("rollout_date")
          .drop_duplicates(subset="country", keep="first")
          .sort_values("rollout_date")
          .reset_index(drop=True)
)

print(f"\nDeduplicated rows: {len(treatment_df)}")
```
